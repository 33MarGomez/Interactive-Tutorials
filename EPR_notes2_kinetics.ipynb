#Empirical Relationships
A good starting point is the parameterization of line broadening to the total nuclear spin of equivalent nuclei. Note the assumption their magnetogyric ratios are positive cancel out first by the square in $\gamma_i$ then by multiplication in $\gamma_{i,j}$ where any negative $M_{I(j)}$ would interact with positive gamma the same way as negative gamma and positive $M_{I(J)}$.
$$\Gamma = \alpha + \sum_{i}\beta_{i}({}^tM_{I(i)})+\sum_{i}\gamma_{i}({}^tM_{I(i)}^2)+\sum_{i<j}\gamma_{i,j}({}^tM_{I(i)}{}^tM_{I(j)})$$

A general empirical relationship is collected by Weil & Bolton for mole fractions $f$ of splitting constants $a$ following interconversion between species a and b [1],[2]. The line is only ever wider than the average of the two exchanging species.
$$\Gamma = \bar{\Gamma} + f_af_b\tau|\bar{\gamma}|(a_a-a_b)^2M_I^2$$
$$\tau = \frac{\tau_a + \tau_b}{\tau_a \tau_b}$$
$$\bar{\Gamma}=f_a\Gamma_{0a}+f_b\Gamma_{0b}$$
$$\bar{\gamma}=f_a\gamma_{a}+f_b\gamma_{b}$$
At very low temperatures, phonon-assisted non-radiative processes dominate. They use a high-temperature approximation in nearly all experimental conditions, and differ by the spin number of the system.
$$\tau_{1,S=1/2} \propto B^{-4}T^{-1}$$
$$\tau_{1,S>1/2} \propto B^{-2}T^{-1}$$

##Susceptibilities and Power
The Steady-state solutions to the Bloch equation in a rotating frame with a perpendicular oscillating field, and steady applied field are used in magnetic susceptibilities. $\omega$ is the oscillation of the applied field with x and y components.
$$M_{x\phi}=-M_z^0\frac{\gamma_eB_1(\omega_B-\omega)\tau_2^2}{1+(\omega_B-\omega)^2\tau_2^2+\gamma_e^2B_1^2\tau_1\tau_2}$$
$$M_{y\phi}=+M_z^0\frac{\gamma_eB_1\tau_2}{1+(\omega_B-\omega)^2\tau_2^2+\gamma_e^2B_1^2\tau_1\tau_2}$$
$$M_{z}=+M_z^0\frac{1+(\omega_B-\omega)^2\tau_2^2}{1+(\omega_B-\omega)^2\tau_2^2+\gamma_e^2B_1^2\tau_1\tau_2}$$
$$\omega_B = -\gamma_e B$$

$$\chi_0=\frac{\kappa\mu_0N_Vg^2\mu_{B(e)}^2}{4k_bT}$$

$$M_Z^0 = \frac{\chi^0 B}{\kappa \mu_0}$$

$M_Z^0$ can also be predicted in terms of steady state population transitions per unit time, $W$, though these quantities might be difficult to extract in practice.
$$M_Z^0 = \frac{1}{2}g\mu_{B(e)}\Delta N_{ss}/V$$
$$\Delta N_{ss} = N\frac{A_{ul}+W_{\downarrow}-W_{\uparrow}}{A_{ul}+2B_{ul}\rho_{\nu}+W_{\downarrow}+W_{\uparrow}}$$

$$B_{ul} = B_{lu}$$
Susceptibilities correspond to x and y magnetizations reduced over the applied field and multiplied by permeability of the medium, with y related to x by the $\tau_2$ constant.
$$\chi'=\chi_0\frac{\omega_B(\omega_B-\omega)\tau_2^2}{1+(\omega_B-\omega)^2\tau_2^2+\gamma_e^2B_1^2\tau_1\tau_2}$$

$$\chi'=\chi_0\frac{\omega_B\tau_2}{1+(\omega_B-\omega)^2\tau_2^2+\gamma_e^2B_1^2\tau_1\tau_2}$$

$$\chi''=\chi'(\omega_B-\omega)\tau_2$$

$$\chi' = +\frac{\kappa\mu_0 M_{x\phi}}{B_1}$$
$$\chi'' = -\frac{\kappa\mu_0 M_{y\phi}}{B_1}$$

The power absorbed by the sample is dictated by susceptability. If $\gamma_e^2B_1^2\tau_1\tau_2 << 1$, that term can be neglected and the power absorption is proportional to $B_1^2$. V is the volume of the sample. The Lorentzian has a half width at its half height predicted by function $\Gamma$.
$$P(\omega)=\frac{\omega\chi''B_1^2}{\mu_0}V$$
$$P=\frac{\pi}{\mu_0}\frac{B_1^2}{(1+\gamma_e^2B_1^2\tau_1\tau_2)^{1/2}}\omega\omega_B\chi_0Y(\omega-\omega_B)$$

$$\Gamma = \frac{1}{\gamma_e\tau_2}(1+\gamma_e^2B_1^2\tau_1\tau_2)^{1/2}$$
$$Y(\omega-\omega_B) = Y_{max} \frac{\Gamma^2}{\Gamma^2+(\omega-\omega_{B})^2}$$
$$Y_{max} = Y\big|_{\omega=\omega_B}=\frac{1}{\pi\Gamma}$$

Defining reaction rates, the complex transverse magnetization is compared to first-rate kinetics. $\epsilon$ is for species a or b as the starting expressions are identical, the stand-in for mole fraction is the magnetization vector such that flux creating species uses rate constant labelled of the opposite species and its magnetization vector. The coefficients $\alpha$ are complex values, with the imaginary portion of the equation dictating intensity of transition.

$$M_{+\phi}=M_{x\phi} + iM_{y\phi}$$
$$\frac{dM_{+\phi}}{dt}+(\tau_2^{-1}-i(\omega_B-\omega))M_{+\phi}=i\gamma_eB_1M_z$$
$$\frac{dM_{+\phi,\epsilon}}{dt}+(\tau_{2,\epsilon}^{-1}-i(\omega_{B,\epsilon}-\omega))M_{+\phi,\epsilon}=i\gamma_eB_1M_{z,\epsilon}+k_{-1 \epsilon}M_{+\phi,\ne \epsilon}-k_{1,\epsilon}M_{+\phi,\epsilon}$$

$$\frac{dM_{+\phi,\epsilon}}{dt} = \frac{dM_{z,\epsilon}}{dt}=0$$

$$M_{z,\epsilon} = f_{\epsilon}\gamma_{\epsilon}M_z^0/\bar{\gamma}$$

$$M_{+\phi}= iB_1M_z^0\frac{f_a\gamma_a(\alpha_b + k_a + k_b) + f_b\gamma_b(\alpha_a + k_a + k_b)}{(\alpha_a+k_a)(\alpha_b+k_b)-k_ak_b}$$

$$M_{+\phi}= iB_1M_z^0\frac{\bar{\gamma}+\tau(f_a\gamma_a\alpha_b+f_b\gamma_b\alpha_a)}{\tau\alpha_a\alpha_b + f_a\alpha_a + f_b\alpha_b}$$

$$\alpha_{\epsilon} = \tau_{2,\epsilon}^{-1}-i(\omega_{B,\epsilon}-\omega)$$

The more compact version involves dividing reciprocal $\tau^{-1} = \tau_a^{-1} + \tau_b^{-1}$ equals $\tau^{-1} = \frac{\tau_b + \tau_a}{\tau_a\tau_b}$ from the equation. The denominator then creates mole fraction $f_{\epsilon} = \tau_{\epsilon}/(\tau_{\epsilon} + \tau_{\ne \epsilon})$ via $k_{\epsilon} = \tau_{\epsilon}^{-1}$. Or apply $\tau/\tau$ with these relationships. At intermediate dynamics, the field shifts. The coalescence point expression is given. In slow dynamics, one of the magnetization vectors is approximated to zero. In fast dynamics, all $\tau$ containing terms are neglected.
$$(B_b - B_a)_t = ((B_b - B_a)_0^2-2(\bar{\gamma}\tau)^{-2})^{1/2}$$
$$\tau=\frac{2\sqrt{2}}{|\bar{\gamma}|(B_b-B_a)}$$
In cases of single electron transfer, a two-species relaxation time is used of radical concentration $[R]$. In the line-width equation, a 2/3 multiplier, 1/3 of encounters produce no reaction, causes the two-species $|2\tau|^{-2}$ cancellation.

$$\Gamma = \Gamma_0 + |3\gamma_e\tau|^{-1}$$

$$k_{(2)}=\frac{1}{2\tau[R]}$$

The imaginary part of the magnetization vectors, $\tau$ dependent, may assemble an Arhennius plot, $\tau^{-1}$ versus $T^{-1}$ as directly corresponding to conversion rate. Relative areas of the absorption curves adequately represent mole fractions [1].

##Pulses

Of value, its approach to the Rabi problem for upper and lower state populations - specifically, the probability of the system being found in upper and lower states. Thus, when the variance in the excitation field matches the fundamental frequency of the electron transition, the system has sinusoidal behaviour, $\omega_B = \gamma_eB$

$$P=\bigg(\frac{\nu_1}{(\nu_1^2+(\nu-\nu_B)^2)^{1/2}}\bigg)^2sin^2(\pi(\nu_1^2+(\nu-\nu_B)^2)^{1/2}(t-t_0))$$

$$\nu_B = g\mu_{B(e)}B/h$$
