##Operators and Selection Rules
This work on operators is useful because it allows the construction of any spin system at will, regardless of its relationship to the Pauli matrices. In fact, one might note the Pauli spin matrices to be a special case of these operators. The only necessary operators are angular momentum operators X, Y, and the ladder operators. Symbol $\hat{\ell}$ is used as a stand in for any of nuclear spin operator $\hat{I}$, electron spin $\hat{S}$, or orbital momentum operator $\hat{L}$. $\hat{\ell}$ is an operator with quantum number $\ell$ as its eigenvalue, and has vector magnitude $(\ell(\ell+1))^{(1/2)}$. It is manipulated as a dot product with itself, $\ell^2 = \ell^T \ell$ in select contexts. $\hat{J}$ is a common operator in experimental work for coupled system $L + S$, which is significant in large-atom hamiltonians. The asterisk means complex conjugate.
$$\ell_+ = \ell_x + i\ell_y$$
$$\ell_- = \ell_x - i\ell_y$$
$$(\ell_+)^{*} = \ell_-$$
$$\ell_x = \frac{1}{2}(\ell_+ + \ell_-)$$
$$\ell_y = \frac{i}{2}(\ell_+ - \ell_-)$$
The ladder operators are simple circular motion senses and $i$ only constrains x and y to be orthogonal in the commutator relationships that define the basis. Otherwise, operators acting on y could rotate the system and not the particle. Of course, $\ell_z$ has an obvious eigenvalue relationship in this basis.
$$\ell_z|m\rangle = m|m\rangle$$
The eigenvalue relationships of senses $+1$ and $-1$ senses simplify expressions. $\ell$ is a square-root of the relationship $\ell^2 = \ell^T\ell = \ell_x\ ^2 + \ell_y\ ^2 + \ell_z\ ^2$ which is recognized as a dot product that gives a single value.

$$\ell_-|\ell,m\rangle= (\ell(\ell+1)-m(m-1))^{(1/2)}|\ell,m-1\rangle$$
